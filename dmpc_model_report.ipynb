# Динамическая модель и варианты управления

## Совместное управление рулением и продольным движением

Я составила плоскую нелинейную модель тягача с полуприцепом. В ней учитываются продольное и боковое движение, вращение тягача и полуприцепа, геометрия сцепки и боковые силы шин с ограничением по сцеплению.

### Геометрические параметры

В модели используются следующие обозначения:

| Обозначение | Смысл                                                                                 |
| ----------- | ------------------------------------------------------------------------------------- |
| $C_1$       | Центр масс тягача                                                                     |
| $C_2$       | Центр масс полуприцепа                                                                |
| $H$         | Точка сцепки                                                                          |
| $l_1$       | Расстояние от центра масс тягача до передней оси                                      |
| $l_2$       | Расстояние от центра масс тягача до сцепки                                            |
| $l_3$       | Расстояние от центра масс тягача до задней оси                                        |
| $b_1$       | Расстояние от сцепки до центра масс полуприцепа                                       |
| $b_3$       | Расстояние от центра масс полуприцепа до задней оси                                   |
| $m_1,m_2$   | Массы тягача и полуприцепа                                                            |
| $I_1,I_2$   | Моменты инерции тягача и полуприцепа относительно вертикальных осей через центры масс |

Ось $x_1$ направлена вперёд вдоль тягача, ось $x_2$ — вперёд вдоль полуприцепа. Положительные углы $\psi_1,\psi_2$ отсчитываются против часовой стрелки.

Положение центра масс полуприцепа выражается через положение центра масс тягача:

$$
p_{C_2}=p_{C_1}-l_2e_1-b_1e_2,
\qquad
e_i=
\begin{bmatrix}
\cos\psi_i\\
\sin\psi_i
\end{bmatrix}.
$$

Расстояние от сцепки до задней оси полуприцепа равно $b_1+b_3$.


### Вектор состояния:

$$
s=
\begin{bmatrix}
X & Y & \psi_1 & \psi_2 &
v_x & v_y & r_1 & r_2 & \delta
\end{bmatrix}^{T}.
$$

Здесь $X,Y$ — положение центра масс тягача, $\psi_1,\psi_2$ — углы тягача и полуприцепа, $v_x,v_y$ — продольная и боковая скорости тягача, $r_1,r_2$ — угловые скорости, $\delta$ — угол передних колёс.

### Вектор управления

Управление включает скорость изменения угла передних колёс и команду продольного ускорения:

$$
\boxed{
u=
\begin{bmatrix}
\dot\delta\\
a_{\mathrm{cmd}}
\end{bmatrix}.
}
$$

Здесь:

- $\dot\delta$ — скорость изменения угла передних колёс;
- $a_{\mathrm{cmd}}$ — команда продольного ускорения, определяющая требуемое тяговое или тормозное усилие.

Угол руля $\delta$ остаётся состоянием модели:

$$
\dot\delta=u_\delta.
$$

Это позволяет ограничивать как сам угол $\delta$, так и скорость его изменения $\dot\delta$.

Команда $a_{\mathrm{cmd}}$ преобразуется в требуемую продольную силу

$$
F_x^{cmd}\approx(m_1+m_2)a_{\mathrm{cmd}},
$$

после чего сила распределяется между осями и ограничивается доступным сцеплением шин. Фактическое ускорение автомобиля определяется полной динамикой

$$
M(\eta)\dot\nu+k(\nu,\eta)=Q.
$$

### Динамика

Введём

$$
\nu=
\begin{bmatrix}
v_x & v_y & r_1 & r_2
\end{bmatrix}^{T},
\qquad
\eta=\psi_1-\psi_2.
$$

Уравнения движения записаны в виде

$$
\boxed{M(\eta)\dot\nu+k(\nu,\eta)=Q}.
$$

Здесь $M$ — матрица инерции, $k$ — скоростные члены, возникающие из-за вращения связанных систем координат и движения полуприцепа, $Q$ — обобщённые силы и моменты.

Матрица инерции получена из кинетической энергии тягача и полуприцепа:

$$
T=
\frac12m_1(v_x^2+v_y^2)
+\frac12I_1r_1^2
+\frac12m_2(v_{2x}^2+v_{2y}^2)
+\frac12I_2r_2^2.
$$

Скорости центра масс полуприцепа выражаются через скорости тягача и геометрию сцепки:

$$
v_{2x}=v_x\cos\eta-(v_y-l_2r_1)\sin\eta,
$$

$$
v_{2y}=v_x\sin\eta+(v_y-l_2r_1)\cos\eta-b_1r_2.
$$

Здесь $l_2$ — расстояние от центра масс тягача до сцепки, $b_1$ — расстояние от сцепки до центра масс полуприцепа.

После подстановки скоростей в кинетическую энергию получаем

$$
\boxed{
M=
\begin{bmatrix}
m_1+m_2 & 0 & 0 & -m_2b_1\sin\eta\\
0 & m_1+m_2 & -m_2l_2 & -m_2b_1\cos\eta\\
0 & -m_2l_2 & I_1+m_2l_2^2 & m_2l_2b_1\cos\eta\\
-m_2b_1\sin\eta & -m_2b_1\cos\eta & m_2l_2b_1\cos\eta & I_2+m_2b_1^2
\end{bmatrix}
}.
$$

Скоростной вектор:

$$
\boxed{
k=
\begin{bmatrix}
m_2l_2r_1^2+m_2b_1r_2^2\cos\eta-(m_1+m_2)r_1v_y\\
(m_1+m_2)r_1v_x-m_2b_1r_2^2\sin\eta\\
-m_2l_2(r_1v_x-b_1r_2^2\sin\eta)\\
-m_2b_1r_1\left[v_x\cos\eta+(l_2r_1-v_y)\sin\eta\right]
\end{bmatrix}
}.
$$

# Формирование продольных и поперечных сил

На каждой оси автомобиля действуют две компоненты силы шины:

- $F_x$ — продольная сила;
- $F_y$ — поперечная сила.

Поэтому для трёх рассматриваемых осей вводятся:

$$
\boxed{
\mathbf F_f^{wheel}
=
\begin{bmatrix}
F_{xf}\\
F_{yf}
\end{bmatrix},
\qquad
\mathbf F_r^{(1)}
=
\begin{bmatrix}
F_{xr}\\
F_{yr}
\end{bmatrix},
\qquad
\mathbf F_t^{(2)}
=
\begin{bmatrix}
F_{xt}\\
F_{yt}
\end{bmatrix}.
}
$$

Здесь:

- $f$ — передняя ось тягача;
- $r$ — задняя ось тягача;
- $t$ — ось полуприцепа.

Продольные силы $F_{xi}$ определяются командой газа/тормоза.

Поперечные силы $F_{yi}$ определяются моделью шин через углы увода.


# 1. Передняя ось тягача

## Продольная и поперечная силы

В системе координат переднего колеса:

$$
\boxed{
\mathbf F_f^{wheel}
=
\begin{bmatrix}
F_{xf}\\
F_{yf}
\end{bmatrix}.
}
$$

Переднее колесо повёрнуто относительно тягача на угол $\delta$, поэтому силу необходимо перевести в систему координат тягача:

$$
\mathbf F_f^{(1)}
=
R(\delta)
\mathbf F_f^{wheel},
$$

где

$$
R(\delta)
=
\begin{bmatrix}
\cos\delta & -\sin\delta\\
\sin\delta & \cos\delta
\end{bmatrix}.
$$

Следовательно,

$$
\boxed{
\mathbf F_f^{(1)}
=
\begin{bmatrix}
F_{xf}\cos\delta-F_{yf}\sin\delta\\
F_{xf}\sin\delta+F_{yf}\cos\delta
\end{bmatrix}.
}
$$

Обозначим

$$
\boxed{
X_f=F_{xf}\cos\delta-F_{yf}\sin\delta
}
$$

$$
\boxed{
Y_f=F_{xf}\sin\delta+F_{yf}\cos\delta.
}
$$


## Скорость передней оси

Передняя ось находится относительно центра масс тягача в точке

$$
\mathbf r_{C_1f}
=
\begin{bmatrix}
l_1\\
0
\end{bmatrix}.
$$

Используем кинематическую формулу твёрдого тела:

$$
\mathbf v_P
=
\mathbf v_C+
\boldsymbol\omega\times\mathbf r_{CP}.
$$

Поэтому

$$
\boxed{
\mathbf v_f
=
\begin{bmatrix}
v_x\\
v_y+l_1r_1
\end{bmatrix}.
}
$$


## Обобщённая сила передней оси

Используем виртуальную мощность:

$$
P_f
=
\left(\mathbf F_f^{(1)}\right)^T
\mathbf v_f.
$$

Следовательно,

$$
P_f
=
X_fv_x+
Y_f(v_y+l_1r_1).
$$

Раскрываем:

$$
P_f
=
X_fv_x+
Y_fv_y+
l_1Y_fr_1.
$$

Сравнивая с

$$
P_f
=
Q_{f,1}v_x+
Q_{f,2}v_y+
Q_{f,3}r_1+
Q_{f,4}r_2,
$$

получаем

$$
\boxed{
Q_f=
\begin{bmatrix}
X_f\\
Y_f\\
l_1Y_f\\
0
\end{bmatrix}.
}
$$

То есть

$$
\boxed{
Q_f=
\begin{bmatrix}
F_{xf}\cos\delta-F_{yf}\sin\delta\\
F_{xf}\sin\delta+F_{yf}\cos\delta\\
l_1(F_{xf}\sin\delta+F_{yf}\cos\delta)\\
0
\end{bmatrix}.
}
$$


# 2. Задняя ось тягача

Задняя ось не повёрнута относительно тягача, поэтому её продольная и поперечная силы уже заданы в системе тягача:

$$
\boxed{
\mathbf F_r^{(1)}
=
\begin{bmatrix}
F_{xr}\\
F_{yr}
\end{bmatrix}.
}
$$


## Скорость задней оси

Положение оси относительно центра масс:

$$
\mathbf r_{C_1r}
=
\begin{bmatrix}
-l_3\\
0
\end{bmatrix}.
$$

Следовательно,

$$
\boxed{
\mathbf v_r=
\begin{bmatrix}
v_x\\
v_y-l_3r_1
\end{bmatrix}.
}
$$


## Обобщённая сила задней оси

Мощность:

$$
P_r
=
F_{xr}v_x+
F_{yr}(v_y-l_3r_1).
$$

То есть

$$
P_r
=
F_{xr}v_x+
F_{yr}v_y-
l_3F_{yr}r_1.
$$

Следовательно,

$$
\boxed{
Q_r=
\begin{bmatrix}
F_{xr}\\
F_{yr}\\
-l_3F_{yr}\\
0
\end{bmatrix}.
}
$$

Продольная сила $F_{xr}$ не создаёт yaw-момента относительно центра масс тягача, поскольку направлена вдоль продольной оси тягача.


# 3. Ось полуприцепа

На оси полуприцепа также одновременно действуют продольная и поперечная силы:

$$
\boxed{
\mathbf F_t^{(2)}
=
\begin{bmatrix}
F_{xt}\\
F_{yt}
\end{bmatrix}.
}
$$

Они заданы в собственной системе координат полуприцепа.


## Скорость сцепки

Сцепка находится относительно центра масс тягача в точке

$$
\mathbf r_{C_1H}
=
\begin{bmatrix}
-l_2\\
0
\end{bmatrix}.
$$

Поэтому

$$
\boxed{
\mathbf v_H=
\begin{bmatrix}
v_x\\
v_y-l_2r_1
\end{bmatrix}.
}
$$


## Скорость оси полуприцепа

Обозначим расстояние от сцепки до оси полуприцепа:

$$
\boxed{
L_t=b_1+b_3.
}
$$

При

$$
\eta=\psi_1-\psi_2
$$

продольное направление полуприцепа в системе тягача:

$$
e_{x_2}
=
\begin{bmatrix}
\cos\eta\\
-\sin\eta
\end{bmatrix}.
$$

Поэтому положение оси относительно сцепки:

$$
\mathbf r_{Ht}
=
-L_te_{x_2}
=
\begin{bmatrix}
-L_t\cos\eta\\
L_t\sin\eta
\end{bmatrix}.
$$

Используем

$$
\mathbf v_t
=
\mathbf v_H+
\boldsymbol\omega_2\times\mathbf r_{Ht}.
$$

Получаем

$$
\boxed{
\mathbf v_t=
\begin{bmatrix}
v_x-L_tr_2\sin\eta\\
v_y-l_2r_1-L_tr_2\cos\eta
\end{bmatrix}.
}
$$


## Перевод продольной и поперечной силы полуприцепа

Сила в системе полуприцепа:

$$
\mathbf F_t^{(2)}
=
\begin{bmatrix}
F_{xt}\\
F_{yt}
\end{bmatrix}.
$$

В системе тягача:

$$
\mathbf F_t^{(1)}
=
\begin{bmatrix}
\cos\eta & \sin\eta\\
-\sin\eta & \cos\eta
\end{bmatrix}
\begin{bmatrix}
F_{xt}\\
F_{yt}
\end{bmatrix}.
$$

Поэтому

$$
\boxed{
X_t=
F_{xt}\cos\eta+
F_{yt}\sin\eta
}
$$

$$
\boxed{
Y_t=
-F_{xt}\sin\eta+
F_{yt}\cos\eta.
}
$$


## Обобщённая сила оси полуприцепа

Виртуальная мощность:

$$
P_t
=
X_t(v_x-L_tr_2\sin\eta)
+
Y_t(v_y-l_2r_1-L_tr_2\cos\eta).
$$

Раскрываем:

$$
P_t
=
X_tv_x+
Y_tv_y-
l_2Y_tr_1
-
L_t
(X_t\sin\eta+Y_t\cos\eta)r_2.
$$

Следовательно,

$$
Q_t=
\begin{bmatrix}
X_t\\
Y_t\\
-l_2Y_t\\
-L_t(X_t\sin\eta+Y_t\cos\eta)
\end{bmatrix}.
$$

При этом

$$
X_t\sin\eta+Y_t\cos\eta
=
F_{yt}.
$$

Поэтому

$$
\boxed{
Q_t=
\begin{bmatrix}
F_{xt}\cos\eta+F_{yt}\sin\eta\\
-F_{xt}\sin\eta+F_{yt}\cos\eta\\
-l_2(-F_{xt}\sin\eta+F_{yt}\cos\eta)\\
-(b_1+b_3)F_{yt}
\end{bmatrix}.
}
$$


# 4. Полный вектор продольных и поперечных сил

Полный вектор обобщённых сил получается суммированием вкладов всех осей:

$$
\boxed{
Q=Q_f+Q_r+Q_t.
}
$$

Следовательно,

$$
\boxed{
Q=
\begin{bmatrix}
F_{xf}\cos\delta-F_{yf}\sin\delta
+F_{xr}
+F_{xt}\cos\eta+F_{yt}\sin\eta
\\[3mm]
F_{xf}\sin\delta+F_{yf}\cos\delta
+F_{yr}
-F_{xt}\sin\eta+F_{yt}\cos\eta
\\[3mm]
l_1(F_{xf}\sin\delta+F_{yf}\cos\delta)
-l_3F_{yr}
-l_2(-F_{xt}\sin\eta+F_{yt}\cos\eta)
\\[3mm]
-(b_1+b_3)F_{yt}
\end{bmatrix}.
}
$$


# 5. Расчет поперечных сил

Поперечные силы определяются через углы увода шин.

Для передней оси:

$$
\boxed{
\alpha_f
=
\operatorname{atan2}
(v_y+l_1r_1,v_x)
-\delta.
}
$$

Для задней оси:

$$
\boxed{
\alpha_r
=
\operatorname{atan2}
(v_y-l_3r_1,v_x).
}
$$

Для полуприцепа сначала скорость его оси переводится в систему полуприцепа:

$$
\begin{bmatrix}
v_{tx}^{(2)}\\
v_{ty}^{(2)}
\end{bmatrix}
=
\begin{bmatrix}
\cos\eta & -\sin\eta\\
\sin\eta & \cos\eta
\end{bmatrix}
\begin{bmatrix}
v_x-L_tr_2\sin\eta\\
v_y-l_2r_1-L_tr_2\cos\eta
\end{bmatrix}.
$$

После этого

$$
\boxed{
\alpha_t=
\operatorname{atan2}
(v_{ty}^{(2)},v_{tx}^{(2)}).
}
$$

### Линейные поперечные силы до ограничения сцеплением

Для каждой оси сначала вычисляется поперечная сила по линейной модели шины:

$$
\boxed{
F_{yf}^{lin}=-C_{\alpha f}\alpha_f,
}
$$

$$
\boxed{
F_{yr}^{lin}=-C_{\alpha r}\alpha_r,
}
$$

$$
\boxed{
F_{yt}^{lin}=-C_{\alpha t}\alpha_t.
}
$$

где $C_{\alpha f}$, $C_{\alpha r}$, $C_{\alpha t}$ — коэффициенты боковой жесткости шин соответствующих осей, а $\alpha_f,\alpha_r,\alpha_t$ — углы увода.

Здесь индекс $lin$ означает силу, полученную из линейной модели до учета предела сцепления с дорогой.

---

### Статические нормальные реакции

На первом этапе используются постоянные статические нормальные реакции $N_f$, $N_r$, $N_t$.

Для полуприцепа рассматриваются вертикальные силы:

- $m_2g$ — сила тяжести полуприцепа;
- $N_t$ — нормальная реакция задней оси полуприцепа;
- $N_H$ — вертикальная реакция в точке сцепки $H$.

Из баланса вертикальных сил:

$$
N_H+N_t=m_2g.
$$

Расстояние от сцепки $H$ до центра масс полуприцепа равно $b_1$, а от центра масс до задней оси — $b_3$. Поэтому расстояние от сцепки до задней оси равно

$$
b_1+b_3.
$$

Из баланса моментов относительно точки сцепки:

$$
N_t(b_1+b_3)-m_2gb_1=0.
$$

Следовательно,

$$
\boxed{
N_t=m_2g\frac{b_1}{b_1+b_3}.
}
$$

Тогда нагрузка, передаваемая через сцепку на тягач:

$$
\boxed{
N_H=m_2g-N_t
     =m_2g\frac{b_3}{b_1+b_3}.
}
$$

На тягач вертикальная сила $N_H$ действует вниз.

Для тягача выполняется баланс вертикальных сил:

$$
N_f+N_r=m_1g+N_H.
$$

Расстояние от центра масс тягача $C_1$ до передней оси равно $l_1$, до задней оси — $l_3$, а до точки сцепки $H$ — $l_2$.

Из баланса моментов относительно центра масс тягача:

$$
-l_1N_f+l_3N_r-l_2N_H=0.
$$

Совместно с балансом вертикальных сил получаем:

$$
\boxed{
N_f=
\frac{
m_1g\,l_3+N_H(l_3-l_2)
}{
l_1+l_3
}.
}
$$

$$
\boxed{
N_r=
\frac{
m_1g\,l_1+N_H(l_1+l_2)
}{
l_1+l_3
}.
}
$$

Таким образом,

$$
\boxed{
N_f+N_r+N_t=(m_1+m_2)g.
}
$$

Последнее равенство используется как проверка корректности статического распределения нагрузки.

---

### Ограничение сцепления

После вычисления линейных поперечных сил для каждой оси независимо применяется ограничение сцепления.

Для передней оси тягача:

$$
\boxed{
F_{xf}^2+F_{yf}^2\leq(\mu_fN_f)^2.
}
$$

Для задней оси тягача:

$$
\boxed{
F_{xr}^2+F_{yr}^2\leq(\mu_rN_r)^2.
}
$$

Для оси полуприцепа:

$$
\boxed{
F_{xt}^2+F_{yt}^2\leq(\mu_tN_t)^2.
}
$$

Таким образом, каждая ось имеет собственный предел результирующей силы сцепления:

$$
\sqrt{F_{xi}^2+F_{yi}^2}\leq\mu_iN_i,
\qquad i\in\{f,r,t\}.
$$

Ограничения сцепления отдельных осей независимы, однако создаваемые ими силы совместно определяют динамику всего тягача с полуприцепом.


# 6. Расчет продольных сил

Командная суммарная продольная сила:

$$
\boxed{
F_x^{cmd}
=
(m_1+m_2)a_{\mathrm{cmd}}.
}
$$

При разгоне

$$
a_{\mathrm{cmd}}>0
$$

ведущей является только задняя ось тягача:

$$
\boxed{
F_{xf}^{cmd}=0,
\qquad
F_{xr}^{cmd}=F_x^{cmd},
\qquad
F_{xt}^{cmd}=0.
}
$$

При торможении

$$
a_{\mathrm{cmd}}<0
$$

тормозная сила распределяется между всеми осями:

$$
\boxed{
F_{xf}^{cmd}=\lambda_fF_x^{cmd},
}
$$

$$
\boxed{
F_{xr}^{cmd}=\lambda_rF_x^{cmd},
}
$$

$$
\boxed{
F_{xt}^{cmd}=\lambda_tF_x^{cmd},
}
$$

где

$$
\lambda_f+\lambda_r+\lambda_t=1.
$$

В первом приближении:

$$
\boxed{
\lambda_i=
\frac{N_i}
{N_f+N_r+N_t}.
}
$$


# 7. Совместное ограничение продольной и поперечной силы

Для каждой оси доступны не две независимые силы, а один ограниченный ресурс сцепления:

$$
\boxed{
F_{xi}^2+F_{yi}^2
\leq
(\mu N_i)^2.
}
$$

Для каждой оси имеется требуемый вектор:

$$
\mathbf F_i^{cmd}
=
\begin{bmatrix}
F_{xi}^{cmd}\\
F_{yi}^{nom}
\end{bmatrix}.
$$

Введём

$$
\boxed{
s_i=
\min
\left(
1,
\frac{\mu N_i}
{\sqrt{
(F_{xi}^{cmd})^2+
(F_{yi}^{nom})^2
}}
\right).
}
$$

Тогда фактически реализуемые силы:

$$
\boxed{
F_{xi}=s_iF_{xi}^{cmd},
\qquad
F_{yi}=s_iF_{yi}^{nom}.
}
$$

Эти ограниченные значения

$$
F_{xf},F_{yf},
F_{xr},F_{yr},
F_{xt},F_{yt}
$$

подставляются в итоговый вектор $Q$.
